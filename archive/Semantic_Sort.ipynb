#### generating columns which can be used together to sort the table

```python
def get_multiple_columns_suggestions(columns, mod='3.5-turbo'):

  '''
    given column names, determine which can be used together to semantically sort the table
  '''

  prompt = f"Given the columns {columns}, determine which columns can be used together to do semantic sorting of the dataset. Focus on columns that have hierarchical, contextual, or categorical relationships/values. Explicitly exclude columns that are primary keys, contain people's names, unique identifiers, or any other columns that cannot be sorted semantically. For each suggested pair of columns, provide a reasoning for their selection, ensuring they are suitable for semantic sorting and do not include primary keys or identifiers. Provide the answer in the format [column1, column2]. Explanation: reasoning about why these columns can be used together to do semantic sorting. Give as many as 5 suggestions with the most probable first."
  response = openai.ChatCompletion.create(
      model=f"gpt-{mod}",
      messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
      max_tokens=1000,
      n=1,
      stop=None,
      temperature=0.7
  )

  return response.choices[0].message['content']
```

```python
def semantic_sort_multiple_column2(column1, values1, column2, values2, mod='3.5-turbo'):
  prompt = f"Given the columns {column1} with values {values1} and {column2} with values {values2}, generate all unique combinations of values from these columns, and then provide a semantic sorting of these combined values. Explain the sorting criteria for the combined values, including the rationale behind why certain combinations come before others and what world knowledge is considered for this purpose, if any. Try to avoid alphabetical sorting and use context and semantic relationships. Provide the results as accurately and relevantly as possible."
  response = openai.ChatCompletion.create(
      model=f"gpt-{mod}",
      messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
      max_tokens=1000,
      n=1,
      stop=None,
      temperature=0.7
  )

  return response.choices[0].message['content']
```

```python
def semantic_sort_multiple_columns_all_rows(column1, values1, column2, values2, mod='3.5-turbo'):
  '''
  Giving all rows consisting of specific columns and asking to sort them semantically.

  '''
  if len(values1) != len(values2):
        raise ValueError("The length of values1 and values2 must be the same")

  combined_values = [(v1, v2) for v1, v2 in zip(values1, values2)]

  prompt = f"""
    Given two columns '{column1}' with values {values1} and '{column2}' with values {values2}, you have specific row values from these columns. Please sort these rows semantically based on the combined values of both columns. The sorting should be contextual and consider real-world knowledge and relationships rather than simple alphabetical order.
For each row, consider how the values in '{column1}' and '{column2}' interact and influence the ranking. Explain the rationale behind the sorting criteria, detailing why certain rows are prioritized over others. Consider hierarchical, contextual, or categorical relationships, and exclude primary keys, names, or other non-semantically sortable columns. Provide the sorted rows and the reasoning behind their order as accurately and relevantly as possible.
Here are the rows: {combined_values}
"""

    # Make the API call to OpenAI
  response = openai.ChatCompletion.create(
        model=f"gpt-{mod}",
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=1000,
        n=1,
        stop=None,
        temperature=0.7
    )

  return response.choices[0].message['content']
```

```python
# Example for semantic_sort_multiple_columns_all_rows

column1 = "Job Title"
values1 = ["Senior Engineer", "Junior Engineer", "Junior Engineer", "Manager", "Manager"]
column2 = "Location"
values2 = ["California", "New York", "California", "Utah", "California"]

response = semantic_sort_multiple_columns_all_rows(column1, values1, column2, values2, mod='4o')
final = response.split("\n\n")
for suggestion in final:
  print(suggestion)
  print("\n")
```

```python
def identify_joinable_datasets(datasets):

    '''
      here, I only give dataset names and its columns as input and try if gpt is able to give suggestions
    '''

    prompt = f'Given the datasets: {datasets}, identify which datasets can be joined based on semantic relationships. Explain the join keys and why these datasets can be joined. Provide as many join suggestions as possible with the most probable first.'

    response = openai.ChatCompletion.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ]
    )

    result = response['choices'][0]['message']['content'].strip()
    return result

def identify_columns_for_sorting(columns):
    prompt = f'Given the columns {columns}, which columns can be used together to do semantic sorting of the dataset? Focus on columns that have hierarchical, contextual, or categorical relationships and if possible try to ignore alphabetical sorting. Provide the answer in the format [column1, column2]. Explanation: reasoning about why these columns can be used together to do semantic sorting. Give as many suggestions as possible with most probable first.'

    response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ]
    )

    result = response['choices'][0]['message']['content'].strip()
    return result

# these datasets have similar column names, and join columns are clearly visible
data1 = {
    "Employee_ID": [1, 2, 3, 4, 5, 6],
    "Name": ["Fiona Black", "Charlie Lee", "Evan Green", "Bob Johnson", "Dana White", "Alice Smith"],
    "Department": ["Engineering", "Marketing", "Sales", "Engineering", "HR", "Engineering"],
    "State": ["Ohio", "New York", "Illinois", "California", "Florida", "Texas"]
}

data2 = {
    "Employee_ID": [1, 2, 3, 4, 5, 6],
    "Job_Level": ["Junior", "Junior", "Associate", "Associate", "Senior", "Senior"],
    "Salary": ["$65,000", "$70,000", "$85,000", "$90,000", "$115,000", "$120,000"]
}

data3 = {
    "Department_ID": [1, 2, 3, 4],
    "Department": ["Engineering", "Marketing", "Sales", "HR"],
    "Manager": ["Alice Smith", "Bob Johnson", "Dana White", "Charlie Lee"]
}

datasets = [
    {"name": "Employee Information", "columns": ["Employee_ID", "Name", "Department", "State"]},
    {"name": "Employee Roles and Salaries", "columns": ["Employee_ID", "Job_Level", "Salary"]},
    {"name": "Department Information", "columns": ["Department_ID", "Department", "Manager"]}
]

joinable_datasets = identify_joinable_datasets(datasets)
print("Joinable Datasets:")
print(joinable_datasets)
```
